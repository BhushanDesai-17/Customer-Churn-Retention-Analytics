```python
import pandas as pd
```

```python
file = "Customer_Retention_Churn_Analysis_50K_Messy_Dataset.xlsx"

customers = pd.read_excel(file, sheet_name = "Customers")
customers
```

```python
# Droping Duplicates

customers = customers.drop_duplicates()
customers
```

```python
# Cross verifying duplicates

customers.duplicated().sum()
```

```python
customers.isnull().sum()
```

```python
# Handiling NULL values

customers['Gender'] = customers['Gender'].fillna(customers['Gender'].mode()[0])

customers['Country'] = customers['Country'].fillna(customers['Country'].mode()[0])

customers['PaymentMethod'] = customers['PaymentMethod'].fillna(customers['PaymentMethod'].mode()[0])

customers['AvgMonthlyUsageHours'] = customers['AvgMonthlyUsageHours'].fillna(customers['AvgMonthlyUsageHours'].median())

customers['SatisfactionScore'] = customers['SatisfactionScore'].fillna(customers['SatisfactionScore'].median())

customers
```

```python
customers['SignupDate'] = pd.to_datetime(customers['SignupDate'])
customers
```

```python
subscriptions = pd.read_excel(file, sheet_name = "Subscriptions")
subscriptions
```

```python
subscriptions.duplicated().sum()
```

```python
subscriptions.isnull().sum()
```

```python
# Handiling NULL Values

subscriptions['DiscountPct'] = subscriptions['DiscountPct'].fillna(subscriptions['DiscountPct'].median())
subscriptions
```

```python
subscriptions.isnull().sum()
```

```python
subscriptions['StartDate'] = pd.to_datetime(subscriptions['StartDate'])
subscriptions['RenewalDate'] = pd.to_datetime(subscriptions['RenewalDate'])
subscriptions
```

```python
supportTickets = pd.read_excel(file, sheet_name = "SupportTickets")
supportTickets
```

```python
supportTickets.duplicated().sum()
```

```python
supportTickets.isnull().sum()
```

```python
# Handiling NULL values

supportTickets['IssueType'] = supportTickets['IssueType'].fillna(supportTickets['IssueType'].mode()[0])
supportTickets
```

```python
supportTickets.isnull().sum()
```

```python
# Saving all the 3 files into CSV

customers.to_csv('customers_cleaned.csv', index = False)
subscriptions.to_csv('subscriptions_cleaned.csv', index = False)
supportTickets.to_csv('SupportTickets_cleaned.csv', index = False)
```
